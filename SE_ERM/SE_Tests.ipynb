```python
import numpy as np
import numpy.random as rnd
import numpy.linalg as linalg
import math
import sys
```

```python
N = 1000000
Sigma = np.array([[1, 0, 0.64, 0.01],[0, 1, 0.01, 0.16],[0.64, 0.01, 0.64, 0.01],[0.01, 0.16, 0.01, 0.23]])
SigmaInv = linalg.inv(Sigma)
L = linalg.cholesky(Sigma)
SumNormal = np.zeros([2,2])
SumSteins = np.zeros([2,2])

for _ in range(N):
    zw = L @ rnd.normal(0, 1, 4)
    #SteinsExtraTerm = np.einsum("ij,j->i", SigmaInv, np.hstack([zw[0:2], [0,0]]))
    SteinsExtraTerm = np.einsum("ij,j->i", SigmaInv, zw)
    SumNormal += np.array([[3*zw[0]*zw[0], 0],[0, 3*zw[1]*zw[1]]])
    SumSteins += np.einsum("i,j->ij", np.array([zw[0]*zw[0]*zw[0], zw[1]*zw[1]*zw[1]]), SteinsExtraTerm[0:2])

print(SumNormal/N)
print(SumSteins/N)
```
